```python
# notebooks/03_investigacion/01_scraping_investigacion.ipynb

import sys
from pathlib import Path

# Configurar path
project_root = Path.cwd().parent
while not (project_root / 'src').exists() and project_root != project_root.parent:
    project_root = project_root.parent
sys.path.insert(0, str(project_root))

import asyncio
from src.scrapers.investigacion_scraper import InvestigacionScraper

# Ejecutar con 60 segundos de espera para anti-bots
scraper = InvestigacionScraper(headless=False, wait_seconds=60)
profiles = await scraper.scrape_all_profiles()
scraper.save_data(profiles)

# Ver resultados
import pandas as pd
df = pd.read_parquet(project_root / 'data' / 'silver' / 'perfiles_investigadores.parquet')
print(f"Total: {len(df)} investigadores")
df.head(10)
```
