### 🦾 Automatización del Pipeline en este Notebook
Para asegurar una comparación justa y caracterizar las experticias actuales de la Universidad de Cuenca, este cuaderno realiza los siguientes pasos automatizados:
[ Archivo Maestro: DSpace ]
│
├── Pestañas de Artículos ───> (Ya listos) ───┐
│                                             ▼
└── Pestaña de Tesis 🚀 ───> [ Extractor REGEX ] ───> [ Filtro Temporal ]
(Lee Metadatos)          (Solo >= 2020)
│
▼
[ Consolidado Capa Silver ]
(Formato Optimizado Parquet)

1. **Lectura Inteligente:** Se cargan los 31,550 registros históricos de tesis.
2. **Minería de Metadatos (Parseo):** Un algoritmo busca la etiqueta oculta `dc.date.issued` en el texto plano para identificar el año real de sustentación de la tesis (ignorando los años de carga al sistema).
3. **Homogeneización Temporal:** Descarta las tesis antiguas y conserva **únicamente** la producción del **2020 en adelante**, igualando el estándar de las otras hojas.
4. **Almacenamiento de Alta Velocidad (Parquet):** Guarda las tres fuentes limpias en formato Parquet para la fase de cruce de nombres (*Fuzzy Matching*).

````python
# ==========================================
# CELDA 1.5: LIMPIEZA DE NOMBRES DE COLUMNAS
# ==========================================
print("="*80)
print("🧹 LIMPIANDO NOMBRES DE COLUMNAS")
print("="*80)

# Función para limpiar TODOS los nombres de columnas
def limpiar_columnas(df, nombre_df):
    """Elimina espacios al inicio/final y normaliza nombres"""
    columnas_originales = df.columns.tolist()
    
    # Crear diccionario de renombre
    rename_dict = {}
    for col in df.columns:
        col_limpio = col.strip()  # Quitar espacios al inicio y final
        if col != col_limpio:
            rename_dict[col] = col_limpio
            print(f"  ✏️ '{col}' → '{col_limpio}'")
    
    # Aplicar renombre
    df = df.rename(columns=rename_dict)
    
    return df

# Aplicar a todos los dataframes
df_tesis = limpiar_columnas(df_tesis, "TESIS")
df_articulos = limpiar_columnas(df_articulos, "ARTÍCULOS")
df_publicaciones = limpiar_columnas(df_publicaciones, "PUBLICACIONES")

# Verificar
print(f"\n✅ Columnas limpias:")
print(f"   TESIS: {df_tesis.columns.tolist()}")
print(f"   ARTÍCULOS: {df_articulos.columns.tolist()}")
print(f"   PUBLICACIONES: {df_publicaciones.columns.tolist()}")

# Verificar que 'METADATOS' existe ahora
if 'METADATOS' in df_tesis.columns:
    print(f"\n✅ 'METADATOS' encontrado en df_tesis")
else:
    print(f"\n❌ 'METADATOS' NO encontrado. Columnas: {df_tesis.columns.tolist()}")
````

````python
import re

def extractor_anho_inteligente(metadato):
    if pd.isna(metadato):
        return None
    
    texto = str(metadato)
    
    # 🕵️‍♂️ Intento 1: Buscar la etiqueta oficial ideal
    match_issued = re.search(r'dc\.date\.issued:\s*(\d{4})', texto)
    if match_issued:
        return int(match_issued.group(1))
        
    # 🕵️‍♂️ Intento 2: Buscar la etiqueta alternativa de migración (Previous issue date)
    match_previous = re.search(r'Previous\s+issue\s+date:\s*(\d{4})', texto, re.IGNORECASE)
    if match_previous:
        return int(match_previous.group(1))
        
    # 🕵️‍♂️ Intento 3: Buscar la fecha de disponibilidad (Último recurso)
    match_available = re.search(r'dc\.date\.available:\s*(\d{4})', texto)
    if match_available:
        return int(match_available.group(1))
        
    return None
````

````python
print("⏳ Procesando cronología de Tesis histórica...")

# 1. Aplicar el extractor solo al dataframe de tesis
df_tesis['ANHO_EMISION'] = df_tesis['METADATOS'].apply(extractor_anho_inteligente)

# 2. Filtrar para quedarnos con la producción actual (2020 en adelante)
df_tesis_filtrada = df_tesis[df_tesis['ANHO_EMISION'] >= 2020].copy()

print("\n==================================================")
print("📊 CONTROL DE CALIDAD - FILTRADO DE TESIS")
print("==================================================")
print(f"🎓 Total Tesis Históricas (Bronze):      {df_tesis.shape[0]}")
print(f"🎯 Tesis Calificadas (>= 2020 en Silver): {df_tesis_filtrada.shape[0]}")
print(f"📉 Registros antiguos archivados:        {df_tesis.shape[0] - df_tesis_filtrada.shape[0]}")
print("==================================================\n")

# Ver una muestra de cómo quedaron los años extraídos
df_tesis_filtrada[['TITULO TESIS', 'ANHO_EMISION']].head()
````

````python
# ==========================================
# CELDA 5: PARSEO UNIVERSAL Y EXCLUSIÓN DE COLUMNAS REDUNDANTES
# ==========================================
print("="*80)
print("🚀 EJECUTANDO EXTRACCIÓN MAESTRA SANEADA (CAPA SILVER)")
print("="*80)

from utils.dspace_parser import ejecutar_parseo_dspace

# Columnas de negocio de entrada que deseamos conservar de la capa Bronze (Sin METADATOS)
COLUMNAS_BASE_TESIS = ['TITULO TESIS'] 
COLUMNAS_BASE_ART = ['TITULO ARTICULO', 'AUTORES', 'REVISTA']
COLUMNAS_BASE_PUB = ['TITULO PUBLICACION', 'AUTORES', 'EDITORIAL']

# 1. Procesar TESIS
print("⏳ Saneando y Parseando Tesis...")
df_tesis_parseado = ejecutar_parseo_dspace(df_tesis_filtrada['METADATOS'])
# Conservamos solo el título limpio original y concatenamos el parseo completo
df_tesis_base = df_tesis_filtrada[df_tesis_filtrada.columns.intersection(COLUMNAS_BASE_TESIS)].reset_index(drop=True)
df_tesis_silver = pd.concat([df_tesis_base, df_tesis_parseado], axis=1)

# 2. Procesar ARTÍCULOS
print("⏳ Saneando y Parseando Artículos...")
df_art_parseado = ejecutar_parseo_dspace(df_articulos['METADATOS'])
df_art_base = df_articulos[df_articulos.columns.intersection(COLUMNAS_BASE_ART)].reset_index(drop=True)
df_articulos_silver = pd.concat([df_art_base, df_art_parseado], axis=1)

# 3. Procesar PUBLICACIONES
print("⏳ Saneando y Parseando Publicaciones...")
df_pub_parseado = ejecutar_parseo_dspace(df_publicaciones['METADATOS'])
df_pub_base = df_publicaciones[df_publicaciones.columns.intersection(COLUMNAS_BASE_PUB)].reset_index(drop=True)
df_publicaciones_silver = pd.concat([df_pub_base, df_pub_parseado], axis=1)

print("\n==================================================")
print("🏆 ARQUITECTURA SILVER SANEADA Y CONSOLIDADA")
print("==================================================")
print(f"🎓 Tesis en Silver:         {df_tesis_silver.shape[0]} filas, {df_tesis_silver.shape[1]} columnas.")
print(f"📝 Artículos en Silver:     {df_articulos_silver.shape[0]} filas, {df_articulos_silver.shape[1]} columnas.")
print(f"📢 Publicaciones en Silver: {df_publicaciones_silver.shape[0]} filas, {df_publicaciones_silver.shape[1]} columnas.")
print("==================================================\n")
````
