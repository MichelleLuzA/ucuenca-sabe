````python
import re

def extractor_anho_inteligente(metadato):
    if pd.isna(metadato):
        return None
    
    texto = str(metadato)
    
    # 🕵️‍♂️ Intento 1: Buscar la etiqueta oficial ideal
    match_issued = re.search(r'dc\.date\.issued:\s*(\d{4})', texto)
    if match_issued:
        return int(match_issued.group(1))
        
    # 🕵️‍♂️ Intento 2: Buscar la etiqueta alternativa de migración (Previous issue date)
    match_previous = re.search(r'Previous\s+issue\s+date:\s*(\d{4})', texto, re.IGNORECASE)
    if match_previous:
        return int(match_previous.group(1))
        
    # 🕵️‍♂️ Intento 3: Buscar la fecha de disponibilidad (Último recurso)
    match_available = re.search(r'dc\.date\.available:\s*(\d{4})', texto)
    if match_available:
        return int(match_available.group(1))
        
    return None
````

````python
print("⏳ Procesando cronología de Tesis histórica...")

# 1. Aplicar el extractor solo al dataframe de tesis
df_tesis['ANHO_EMISION'] = df_tesis['METADATOS'].apply(extractor_anho_inteligente)

# 2. Filtrar para quedarnos con la producción actual (2020 en adelante)
df_tesis_filtrada = df_tesis[df_tesis['ANHO_EMISION'] >= 2020].copy()

print("\n==================================================")
print("📊 CONTROL DE CALIDAD - FILTRADO DE TESIS")
print("==================================================")
print(f"🎓 Total Tesis Históricas (Bronze):      {df_tesis.shape[0]}")
print(f"🎯 Tesis Calificadas (>= 2020 en Silver): {df_tesis_filtrada.shape[0]}")
print(f"📉 Registros antiguos archivados:        {df_tesis.shape[0] - df_tesis_filtrada.shape[0]}")
print("==================================================\n")

# Ver una muestra de cómo quedaron los años extraídos
df_tesis_filtrada[['TITULO TESIS', 'ANHO_EMISION']].head()
````

## Match Etapa 1: cedula exacta contra el catastro de docentes

DSpace trae, cuando existe, la cedula del autor/director en `dc.ucuenca.idautor` (ya extraida como `ID_AUTOR_N` / `ID_DIRECTOR_TESIS` por `ejecutar_parseo_dspace`). Antes de hacer cualquier fuzzy matching por nombre, cruzamos esa cedula contra `Config.CATRASTRO_DOCENTES` — es exacto, no necesita normalizar nombres, y cubre una parte grande de articulos/publicaciones. El nombre por fuzzy (Etapa 2) queda para lo que no matchea aqui, sobre todo tesis (el director casi nunca trae id).

````python
# ==========================================
# CELDA 5.5: MATCH EXACTO POR CEDULA (ID_AUTOR_N / ID_DIRECTOR_TESIS)
# ==========================================
print("="*80)
print("MATCH ETAPA 1: CEDULA EXACTA CONTRA CATASTRO DE DOCENTES")
print("="*80)


def normalizar_cedula(serie):
    return (serie.astype(str)
                 .str.strip()
                 .str.replace(r'[^0-9]', '', regex=True)
                 .str.zfill(10))


df_catastro = pd.read_parquet(Config.CATRASTRO_DOCENTES)
cols_docente = ['CEDULA_NORM', 'NOMBRES COMPLETOS', 'CAMPO AMPLIO (UNESCO)']
df_catastro_ref = df_catastro[[c for c in cols_docente if c in df_catastro.columns]].drop_duplicates('CEDULA_NORM')
print(f"Catastro de referencia: {len(df_catastro_ref)} docentes (CEDULA_NORM unica)")


def match_por_cedula_simple(df, id_col):
    """Para tesis: una sola cedula candidata (ID_DIRECTOR_TESIS)."""
    df = df.copy()
    if id_col not in df.columns:
        df['CEDULA_MATCH'] = None
        df['METODO_MATCH_DOCENTE'] = None
        return df
    df['CEDULA_MATCH'] = normalizar_cedula(df[id_col].fillna(''))
    df.loc[df[id_col].isna(), 'CEDULA_MATCH'] = None
    en_catastro = df['CEDULA_MATCH'].isin(df_catastro_ref['CEDULA_NORM'])
    df.loc[~en_catastro, 'CEDULA_MATCH'] = None
    df['METODO_MATCH_DOCENTE'] = df['CEDULA_MATCH'].notna().map({True: 'cedula_directa', False: None})
    return df


def match_por_cedula_multiautor(df):
    """Para articulos/publicaciones: cualquiera de los ID_AUTOR_N puede ser el docente."""
    df = df.copy()
    id_cols = [c for c in df.columns if c.startswith('ID_AUTOR_')]
    df['CEDULA_MATCH'] = None
    df['METODO_MATCH_DOCENTE'] = None
    catastro_set = set(df_catastro_ref['CEDULA_NORM'])
    for c in id_cols:
        norm = normalizar_cedula(df[c].fillna(''))
        hit = df[c].notna() & df['CEDULA_MATCH'].isna() & norm.isin(catastro_set)
        df.loc[hit, 'CEDULA_MATCH'] = norm[hit]
        df.loc[hit, 'METODO_MATCH_DOCENTE'] = 'cedula_directa'
    return df


df_tesis_silver = match_por_cedula_simple(df_tesis_silver, 'ID_DIRECTOR_TESIS')
df_articulos_silver = match_por_cedula_multiautor(df_articulos_silver)
df_publicaciones_silver = match_por_cedula_multiautor(df_publicaciones_silver)

# Traer nombre y campo amplio del docente matcheado (util para Power BI y para
# corroborar despues los matches por nombre de la Etapa 2)
df_tesis_silver = df_tesis_silver.merge(df_catastro_ref, how='left', left_on='CEDULA_MATCH', right_on='CEDULA_NORM', suffixes=('', '_DOCENTE'))
df_articulos_silver = df_articulos_silver.merge(df_catastro_ref, how='left', left_on='CEDULA_MATCH', right_on='CEDULA_NORM', suffixes=('', '_DOCENTE'))
df_publicaciones_silver = df_publicaciones_silver.merge(df_catastro_ref, how='left', left_on='CEDULA_MATCH', right_on='CEDULA_NORM', suffixes=('', '_DOCENTE'))

print("\nCobertura Etapa 1 (cedula exacta):")
for nombre, df in [('Tesis', df_tesis_silver), ('Articulos', df_articulos_silver), ('Publicaciones', df_publicaciones_silver)]:
    total = len(df)
    matcheados = df['METODO_MATCH_DOCENTE'].notna().sum()
    print(f"   {nombre}: {matcheados}/{total} ({matcheados/total*100:.0f}%) — quedan {total - matcheados} para la Etapa 2 (fuzzy por nombre)")
````

## Match Etapa 2: fuzzy por nombre + corroboracion por campo amplio

Para lo que no matcheo por cedula (sobre todo directores de tesis, que casi nunca traen id en DSpace): se compara el nombre del registro (formato DSpace 'Apellidos, Nombres') contra `NOMBRES COMPLETOS` del catastro con `token_sort_ratio` (no importa el orden apellido/nombre ni mayusculas/tildes, ya se normalizan antes).

Si el registro trae su propio `UNESCO_AMPLIO` (el campo amplio de ESE articulo/tesis, tomado de los metadatos DSpace) se compara contra el `CAMPO AMPLIO (UNESCO)` del docente candidato: si coinciden, sube la confianza; si contradicen, se descarta el candidato aunque el nombre se parezca mucho. Cuando el registro no trae `UNESCO_AMPLIO` (la mayoria de tesis, ~89%), no hay nada que corroborar y se decide solo por el puntaje de nombre.

IMPORTANTE: el catastro (`Config.CATRASTRO_DOCENTES`) es solo personal **activo hoy**. Un director de tesis que ya no trabaja en la universidad no va a aparecer aqui aunque el nombre matchee perfecto — es la limitacion esperada, no un bug (decision tomada: partir con activos, ampliar despues si hace falta).

````python
# ==========================================
# CELDA 5.6: MATCH FUZZY POR NOMBRE (con corroboracion por campo amplio)
# ==========================================
print("="*80)
print("MATCH ETAPA 2: FUZZY POR NOMBRE + CORROBORACION POR CAMPO AMPLIO")
print("="*80)

import unicodedata
from thefuzz import fuzz, process

UMBRAL_ALTA = 90     # score >= esto y corrobora (o no hay UNESCO_AMPLIO que corroborar) -> aceptado
UMBRAL_REVISAR = 80  # score >= esto y corrobora -> aceptado pero marcado para revision manual


def limpiar_nombre(x):
    """Sin tildes, sin puntuacion, mayusculas, espacios colapsados."""
    if x is None or (isinstance(x, float) and pd.isna(x)):
        return None
    t = unicodedata.normalize('NFKD', str(x)).encode('ascii', 'ignore').decode('ascii')
    t = re.sub(r'[^A-Za-z\s]', ' ', t).upper()
    return re.sub(r'\s+', ' ', t).strip()


df_catastro_ref['NOMBRE_LIMPIO'] = df_catastro_ref['NOMBRES COMPLETOS'].map(limpiar_nombre)
_choices = dict(zip(df_catastro_ref['NOMBRE_LIMPIO'], df_catastro_ref['CEDULA_NORM']))
_choice_list = list(_choices.keys())
_docente_por_cedula = df_catastro_ref.set_index('CEDULA_NORM')


def _mejor_candidato_por_nombre_unico(nombres_unicos):
    """process.extractOne una sola vez por nombre unico (no por fila) para no repetir trabajo."""
    out = {}
    for n in nombres_unicos:
        if not n:
            continue
        res = process.extractOne(n, _choice_list, scorer=fuzz.token_sort_ratio)
        if res:
            match_name, score = res
            out[n] = (score, _choices[match_name])
    return out


def match_fuzzy_nombre(df, name_cols, unesco_col=None):
    """Completa CEDULA_MATCH/METODO_MATCH_DOCENTE solo donde la Etapa 1 no matcheo.
    name_cols: columnas de nombre a probar (para tesis, DIRECTOR_TESIS; para
    articulos/publicaciones, todas las AUTOR_N — cualquiera puede ser el docente).
    """
    df = df.copy()
    pendientes = df['METODO_MATCH_DOCENTE'].isna()

    nombres_limpios = {c: df[c].map(limpiar_nombre) for c in name_cols if c in df.columns}
    todos = set()
    for s in nombres_limpios.values():
        todos |= set(s.dropna().unique())
    lookup = _mejor_candidato_por_nombre_unico(todos)

    best_score = pd.Series(0, index=df.index)
    best_cedula = pd.Series(None, index=df.index, dtype=object)
    for s in nombres_limpios.values():
        score_col = s.map(lambda n: lookup.get(n, (0, None))[0] if n else 0)
        cedula_col = s.map(lambda n: lookup.get(n, (0, None))[1] if n else None)
        mejora = score_col > best_score
        best_score = best_score.where(~mejora, score_col)
        best_cedula = best_cedula.where(~mejora, cedula_col)

    campo_docente = best_cedula.map(
        lambda cd: _docente_por_cedula.loc[cd, 'CAMPO AMPLIO (UNESCO)'] if cd in _docente_por_cedula.index else None)
    if unesco_col and unesco_col in df.columns:
        campo_registro = df[unesco_col]
        coincide = (campo_registro.isna()) | (
            campo_registro.str.strip().str.lower() == campo_docente.fillna('').str.strip().str.lower())
    else:
        coincide = pd.Series(True, index=df.index)

    alta = pendientes & (best_score >= UMBRAL_ALTA) & coincide
    revisar = pendientes & (~alta) & (best_score >= UMBRAL_REVISAR) & coincide

    df.loc[alta, 'CEDULA_MATCH'] = best_cedula[alta]
    df.loc[alta, 'METODO_MATCH_DOCENTE'] = 'fuzzy_nombre'
    df.loc[revisar, 'CEDULA_MATCH'] = best_cedula[revisar]
    df.loc[revisar, 'METODO_MATCH_DOCENTE'] = 'fuzzy_revisar'
    if 'FUZZY_SCORE' not in df.columns:
        df['FUZZY_SCORE'] = None
    df.loc[pendientes, 'FUZZY_SCORE'] = best_score[pendientes]
    return df


df_tesis_silver = match_fuzzy_nombre(df_tesis_silver, ['DIRECTOR_TESIS'], unesco_col='UNESCO_AMPLIO')

autor_cols_art = [c for c in df_articulos_silver.columns if c.startswith('AUTOR_')]
df_articulos_silver = match_fuzzy_nombre(df_articulos_silver, autor_cols_art, unesco_col='UNESCO_AMPLIO')

autor_cols_pub = [c for c in df_publicaciones_silver.columns if c.startswith('AUTOR_')]
df_publicaciones_silver = match_fuzzy_nombre(df_publicaciones_silver, autor_cols_pub, unesco_col='UNESCO_AMPLIO')

# Refrescar nombre/campo del docente para los matches nuevos (fuzzy) que la
# Etapa 1 no habia podido rellenar via merge
def _rellenar_info_docente(df):
    falta_info = df['CEDULA_MATCH'].notna() & df['NOMBRES COMPLETOS'].isna()
    if not falta_info.any():
        return df
    relleno = df.loc[falta_info, 'CEDULA_MATCH'].map(
        lambda cd: _docente_por_cedula.loc[cd] if cd in _docente_por_cedula.index else None)
    for col in ['NOMBRES COMPLETOS', 'CAMPO AMPLIO (UNESCO)']:
        df.loc[falta_info, col] = relleno.map(lambda r: r[col] if r is not None else None)
    return df


df_tesis_silver = _rellenar_info_docente(df_tesis_silver)
df_articulos_silver = _rellenar_info_docente(df_articulos_silver)
df_publicaciones_silver = _rellenar_info_docente(df_publicaciones_silver)

print("\nCobertura final (Etapa 1 cedula + Etapa 2 fuzzy):")
for nombre, df in [('Tesis', df_tesis_silver), ('Articulos', df_articulos_silver), ('Publicaciones', df_publicaciones_silver)]:
    print(f"\n{nombre}:")
    print(df['METODO_MATCH_DOCENTE'].value_counts(dropna=False))
````
